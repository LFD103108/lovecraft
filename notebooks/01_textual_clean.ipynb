# 1. Information from the Texts

We now proceed to calculate some relevant parameters for each text. For now, we shall obtain:

1. Number of sentences;
2. Number of words;
3. Number of Characters.

We store these information in the Pandas DataFrame $\tt df\_texts$, to be saved as a $\tt .csv$ file.

```python
# creating the dataframe
df_texts = pd.DataFrame()          # creates an empty dataframe
df_texts['text'] = biblio.texts.keys()  # adds the name of the stories
df_texts['file'] = biblio.files  # adds the name of the stories

# displays the dataframe
df_texts
```

We now add the relevant information to the DataFrame.

```python
# Adding statistics 

# number of sentences
df_texts['sentence_count'] = [ len(biblio.sentences[text]) for text in df_texts['text'] ]  

# number of words
df_texts['word_count'] = [ len( biblio.texts[text].split() ) for text in df_texts['text'] ]  

# vocabulary, i.e. number of different words; 
# here, a set() object was used to filter out duplicate words
df_texts['vocabulary'] = [ len( set( biblio.texts[text].split() ) ) for text in df_texts['text'] ] 

# number of characters
df_texts['character_count'] = [ len(biblio.texts[text]) for text in df_texts['text'] ]  

# diaplays the updated DataFrame
df_texts.to_csv('data/gold/texts.csv', index=False)  # saves the DataFrame to a CSV file
df_texts.head()
```
