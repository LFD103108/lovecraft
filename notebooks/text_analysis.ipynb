```python
# Libraries

# General
import numpy as np                      # imports the Numpy library for numerical tools
import pandas as pd                     # imports the Pandas library for data manipulation and analysis

# Plotting Options
from matplotlib import pyplot as plt                           # imports the Pyplot module from the Matplotlib library for plotting
plt.rcParams['text.usetex'] = True                             # enables LaTeX rendering for text in plots
plt.rcParams['text.latex.preamble'] = r'\usepackage{amsmath}'  # LaTeX preamble

# File management
import pathlib as Path                  # imports path tools from Pathlib for working with file paths
import os                               # imports the OS library for interacting with the operating system
os.chdir('..')                          # changes to the parent directory

# tools for text processing
from tools import text_tools
```

```python
biblio = text_tools.bibliography()  # defines a bibliography object
print( biblio.files )               # lists the files in the directory
```

```python
# Numerical Representation Plots

# loops over each text
for text in biblio.numerical_dict.keys():

    # plots the Fourier transform
    plt.plot( np.abs(biblio.numerical_dict[text]) )         # plots the text of Call of Cthulhu

    # plot settings
    plt.title(text + ': ' + 'Numerical Representation')  # sets the title of the plot
    plt.xlabel('Character Index')  # sets the label for the x-axis
    plt.ylabel('Character Value')  # sets the label for the y-axis

    # log scale 
    #plt.xscale('log')  # sets the y-axis to a logarithmic scale
    plt.yscale('log')  # sets the y-axis to a logarithmic scale

    # saves the figure
    plt.savefig('plots/numerical_representations/'+ text + '_numerical.png')  
    
    # displays the plots
    plt.show()
```

```python
# Fourier Transform plots

# loops over each text
for text in biblio.numerical_dict.keys():

    # plots the Fourier transform
    plt.plot( np.abs(biblio.fourier_dict[text]) )         # plots the text of Call of Cthulhu

    # plot settings
    plt.title(text + ': ' + 'Fourier Transform')  # sets the title of the plot
    plt.xlabel('Frequency')  # sets the label for the x-axis  
    plt.ylabel('Amplitude')  # sets the label for the y-axis

    # log scale 
    plt.xscale('log')  # sets the y-axis to a logarithmic scale
    plt.yscale('log')  # sets the y-axis to a logarithmic scale

    # saves the figure
    plt.savefig('plots/fourier_transforms/'+ text + '_fourier.png')  
    
    # displays the plots
    plt.show()
```
